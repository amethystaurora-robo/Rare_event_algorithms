```python
import numpy as np
import pandas as pd
import math
import seaborn as sns
from pathlib import Path
```

```python
folder = Path("experiments/degeneracy/amocs/yearly")

files = sorted(folder.glob("*.csv"))

if not files:
    raise FileNotFoundError(f"No CSV files found in {folder}")

dataframes = []

for file in files:
    temp = pd.read_csv(file)

    # Preserve the source filename for traceability
    temp["source_file"] = file.name

    dataframes.append(temp)

all_data = pd.concat(
    dataframes,
    ignore_index=True,
    sort=False
)

print(f"Read {len(files)} files")
print(f"Combined shape: {all_data.shape}")
```

```python
#threshold for collapse is 0 (trajs go from 0.15 to -0.05)
all_data["collapsed"] = all_data["Last AMOC value"].apply(
    lambda x: 0 if x > 0 else 1
)

weights_filtered = all_data.filter(items=["collapsed","Last AMOC value", "Label", "Exponential","Weight","Init Label","Agg Label","Run name"])
filtered=weights_filtered.sort_values('Run name')
rrr = sum(filtered['Exponential']/len(filtered))
#arr.append(rrr)
filtered['R']=rrr
filtered['W_reverse']=1.0/filtered['Exponential']
filtered['R_reverse'] = np.mean(filtered['W_reverse'])
filtered['w_reverse'] = filtered['W_reverse']/filtered['R_reverse']

natural_probability = np.sum(filtered['collapsed']*filtered['w_reverse'])/np.sum(filtered['w_reverse'])
```
